## 00. パタトクカシーー
2つの文字列「パトカー」と「タクシー」の文字を先頭から交互に連結し、文字列「パタトクカシーー」を得よ。

```python
from itertools import zip_longest
x="パトカー"
y="タクシー"

result="".join(
    x+y
    for x, y in zip_longest(x,y)
)

result
```

```python
x="パトカー"
y="タクシー"
result1=""

for i in range(len(x)):
    result1+=x[i]+y[i]

result1
```

## 02. 文字列の逆順
文字列"stressed"の文字を逆に（末尾から先頭に向かって）並べた文字列を得よ。

```python
x="stressed"

result2=x[::-1] #文字列を末尾から先頭に取り出す

result2
```

```python
x="stressed"
result2=""

for i in range(len(x)):
    result2+=x[len(x)-i-1]

result2
```

```python
word = "stressed"
print(word[0:3])   # "str"  0文字目から3文字目の手前まで
print(word[:3] )    # "str"  最初から3文字目の手前まで
print(word[3:])     # "essed" 3文字目から最後まで
print(word[::2])    # "srsd" 1文字おき
print(word[1::2] )  # "tes"  2文字目から1文字おき
print(word[::-1])   # "desserts" 逆順
print(word[-3:] )   # "sed" 最後の3文字
print(word[:-3] )   # "stres" 最後の3文字以外
print(word[-1]  )   # "d" 最後の1文字
```

## 03. 円周率
"Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics."という文を単語に分解し、各単語の（アルファベットの）文字数を先頭から出現順に並べたリストを作成せよ。

```python
text="Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics."

words=text.replace(",","").replace(".","").split()  #replace() で , と . を消してから、split() で単語ごとのリストを作る

result3=[len(word) for word in words]

result3
```

```python
result3=[]

for word in words:
    length=len(word)
    result3.append(length)

result3
```
